## Knuth’s Literate Programming: Programming as Literature
Donald Knuth once said, “A good program should be a work of art like a good book.” He envisioned programs not as mere instructions for computers but as stories crafted for humans, blending code with narrative to reveal the *why* behind every decision. In his world, reading a program is like diving into a novel, with characters (algorithms), plot twists (optimizations), and a narrative arc that makes the logic come alive. For *Magrathea*, this means treating our code as a tale of cosmic creation, where each function and formula contributes to the grandeur of a Voyager-style animation.

Knuth developed tools like WEB to weave code and documentation, but we’re taking a modern approach. Using Jupyter notebooks, we combine English prose, mathematical equations, and executable Python code (powered by *Magrathea*’s library, NumPy, and Cython) to tell this story. The result is a series of notebooks that read like chapters in a book, inviting you to explore the algorithms and animations as if you’re uncovering a mystery by the fireside.

## The Origin of Magrathea

Let’s step back to 1979, when the Voyager spacecraft were hurtling toward Jupiter. Charles Kohlhase, a mission designer at JPL, was hard at work optimizing the flyby trajectories. His code generated wireframe views of planets as seen through Voyager’s cameras—functional, first-person snapshots for science planning. Enter James Blinn, fresh from his computer graphics PhD at the University of Utah and newly hired at JPL. Blinn took Kohlhase’s work to a whole new level, crafting photorealistic animations that brought the Voyager mission to life. Picture texture-mapped spheres for planets and moons, paired with a triangle mesh model of the Voyager spacecraft itself. While Kohlhase’s view was like looking through Voyager’s eyes, Blinn’s was a third-person perspective—a cosmic “over the shoulder” shot, letting us watch Voyager observe the planets in stunning detail.

Blinn’s system didn’t have a formal name, but its output became iconic: those breathtaking Voyager animations that graced the evening news as B-roll during the encounters. In a conversation with a BBC interviewer, someone noted that Blinn’s work sounded like *Magrathea*, the planet-building factory from Douglas Adams’ *The Hitchhiker’s Guide to the Galaxy*. Blinn loved the name and used it in presentations about his system, though the code itself kept its utilitarian Fortran labels, like `SPACE`, constrained by the era’s computing limits.

Blinn worked in what we might call the primeval age of computers, likely on a custom system—possibly a Unix variant—with Fortran’s restrictive short identifiers. As of today, September 2025, James Blinn is still with us and has hinted at releasing his original source code. Until then, he’s shared enough about his system in talks and writings for us to reimagine it from scratch. That’s the heart of this project: *Magrathea*, a modern reimplementation of Blinn’s vision using Python, NumPy, and Cython, wrapped in Jupyter notebooks. These notebooks don’t just recreate the animations; they tell the story of how they’re made, blending code, math, and prose in the spirit of Knuth’s literate programming—a tale as captivating as a fireside novel.

### Parametric Equations: The Heart of Motion

In Algebra, we meet the function graph. We see the line $y=mx+b$, the parabola $y=ax^2+bx+c$, and the general curve $y=\operatorname{f}(x)$. When I first dove into computer graphics via the *Postscript Language Reference Manual*, I met parametric equations -- curves where $x$ and $y$ aren’t directly tied but are both functions of a parameter $t$, which I like to think of as *time*. This parameter drives everything in *Magrathea*’s animations, from the path of a camera to the trajectory of a ray tracing a planet’s surface. Let’s explore why parametrics are a game-changer for our Voyager-style flybys.

The simplest and most powerful parametric form in *Magrathea* is the **ray**, a linear equation used in ray tracing to model light or camera paths. A ray in 3D space is written as a vector equation:

$$\vec{r}(t) = \vec{r}_0 + \vec{v}t$$

Here, $\vec{r}_0$ is the starting point (a 3D vector, like the camera’s position), $vec{v}$ is the direction vector, and $t \geq 0$ moves us along the ray. In component form, for a 2D ray, this might look like:

$$\begin{eqnarray}
x(t) &=& x_0 + v_x t \\
y(t) &=& y_0 + v_y t
\end{eqnarray}$$

This is the backbone of *Magrathea*’s rendering, letting us trace rays from Voyager’s camera to a planet’s surface, computing intersections to create those iconic flyby visuals.

Another example, which sparked my love for parametrics, is the **Bezier curve**. While *Magrathea* doesn’t use them (yet!), they’re a cornerstone of computer graphics, like in Postscript’s `curveto` command. A cubic Bezier curve defines \( x \) and \( y \) as:

$$\begin{eqnarray}
x(t) &=& a_x t^3 + b_x t^2 + c_x t + d_x \\
y(t) &=& a_y t^3 + b_y t^2 + c_y t + d_y
\end{eqnarray}$$
where the coefficients come from control points, shaping smooth curves for things like text outlines or paths in vector graphics. In vector form, this becomes:

$$\vec{r}(t) = \vec{a} t^3 + \vec{b} t^2 + \vec{c} t + \vec{d}$$

where each coefficient is a vector. Bezier curves taught me to see $t$ as a storyteller, guiding points through space over time -- a concept we apply in *Magrathea*’s linear rays.

### Matrices: Transforming space

The other major concept I learned from the Postscript red book is the **Current Transformation Matrix** or transformation matrices in general. At the time, I didn’t fully grasp matrices, but I told a co-worker I’d become “dangerous” once I did. I just used `translate`, `rotate`, non-uniform `scale`, etc. I knew that the current transformation matrix encoded any of these, and somehow could encode the concatenation of all of these, but I didn't know how. Matrices turn out to be the key to transforming objects -- rotating, scaling, and positioning our planets and spacecraft.

Looking back, the seeds of linear algebra were there in Algebra II’s systems of equations and Algebra I’s quadratic formula. I loved solving nonlinear systems by substitution but struggled with the addition method, missing the roots of linear algebra. Similarly, a quadratic formula word problem stuck with me: a diver jumps off a board, and we calculate when she reaches 27 feet (twice, up and down), 28 feet (once, at the apex), or 29 feet (never, due to a negative discriminant). This showed me algebra’s power to model motion—perfect for animating Voyager’s flybys.


### What You Need for Magrathea

The good news? *Magrathea* relies on algebra and linear algebra, not calculus. The concepts you learned in high school Algebra I and II are enough to get started, with linear algebra as the star of the show. Here’s what you need:

- **Vectors**: These define our planet’s vertices or camera positions in 3D space.
- **Linear Transformations (Matrices)**: Matrices rotate, scale, and translate our models, like spinning a planet or moving Voyager’s camera.
- **Basis Vectors**: These set up our coordinate systems, ensuring our animations align with the universe’s geometry.
- **Matrix Transpose and Inverse**: These help us reverse transformations, like adjusting a camera’s view.
- **Determinants**: We use orthogonal matrices (with determinant 1) for clean rotations and scaling in our rendering pipeline.

If these terms feel unfamiliar, don’t worry! The [3Blue1Brown linear algebra series](https://www.youtube.com/playlist?list=PLZHQObOWTQDPD3MizzM2xVFitgF8hE_ab) explains them beautifully. Watch it to build or refresh your intuition, especially if a concept in *Magrathea*’s notebooks feels unclear. With these tools, you’re ready to craft planets and flybys, just like James Blinn did.


### What You Don’t Need: The Pure-Math Power of Python and Numpy

Here’s a refreshing twist: *Magrathea* doesn’t require fancy graphics libraries, GPU programming, or anything beyond Python and NumPy. I’ll let you in on a secret -- I don’t even know how to program a GPU, and you don’t need to either! Our Voyager-style animations, inspired by James Blinn’s iconic work, are built purely on math. With just Python and NumPy’s mathematical muscle, we can craft planets, shade surfaces, and animate flybys, all on a standard laptop.

What makes this possible? NumPy's vectorized operations, which let us perform calculations on entire arrays at once, no loops required. For example, to rotate a spacecraft mesh, we gather all its vertices into a 3xN matrix and transform them with a single matrix multiplication: `v_out = M @ v_in`. This one line rotates thousands of points instantly, expressing the idea of transformation clearly and efficiently. Or consider rendering a triangle mesh: we evaluate the edge function for all pixels in a triangle’s bounding box at once, create a boolean mask, and assign colors in a single operation. It’s fast, concise, and -- dare I say -- beautiful, like a well-crafted sentence in a novel.

If I find myself writing a loop to iterate over a NumPy array, I know I’ve missed an opportunity. There’s almost always a one-line NumPy expression that’s not only faster (thanks to NumPy’s optimized C backend) but also a clearer expression of the concept. In *Magrathea*, this approach lets us focus on the math behind Blinn’s animations—linear algebra, parametric equations, and nothing more. No OpenGL, no DirectX, no shaders. Just pure math, brought to life in Jupyter notebooks, ready for you to run and explore by the fireside.

### Part 1: Core Concepts (`literate_doc/`)

These notebooks lay the mathematical and algorithmic foundation, using Python and NumPy to express ideas clearly, without graphics libraries or GPUs.

- **01 - Setting Up Magrathea**
  Get *Magrathea* running with Python, NumPy, Cython, and PyCharm’s Jupyter support. We’ll walk through installing dependencies and running your first notebook, no GPU required—just pure math.

- **02 - The Frame Buffer and Layering**
  Explore the frame buffer, our canvas for rendering pixels. Learn how we layer objects (planets, stars, spacecraft) using NumPy arrays to composite scenes, echoing Blinn’s approach to animation frames.

- **03 - The Camera Matrix**
  Dive into the camera matrix, built from down, right, and direction vectors. We’ll see how it defines a world-to-camera transformation, using a single matrix multiplication (`v_out = M @ v_in`) to position our viewpoint for Voyager’s flybys.

- **04 - Projection: Why Divide by Z Works**
  Unravel the magic of perspective projection. Why does dividing by \( z \) in camera coordinates create a 2D image? We’ll use linear algebra to project 3D planets onto a 2D screen, just like Blinn did.

- **05 - Parametric Equations and Rays**
  Discover parametric equations, where \( x = f_x(t) \), \( y = f_y(t) \), or in vector form, \( \mathbf{r}(t) = \mathbf{r}_0 + \mathbf{v}t \). Rays are the stars here, tracing light paths for rendering, while Bezier curves offer a glimpse into parametric possibilities.

- **06 - Ray-Surface Intersection**
  Learn to find where rays meet surfaces, solving equations for points on primitives. We’ll start with planes (simple and elegant) and move to ellipsoids, key to modeling planets like Jupiter. Expect vectorized NumPy solutions for speed and clarity.

- **07 - Illumination Models**
  Light up our planets with math! We’ll cover:
  - *Surface Normals*: Vectors for calculating light interactions.
  - *Lambertian Reflectors*: Simple diffuse lighting for matte surfaces.
  - *Modifications*:
    - Oren-Nayar for rough astronomical surfaces.
    - Opposition effect for bright spots on moons.
    - Phong for shiny, metallic looks.
    - POV-Ray’s `brilliance` for unique highlights.
  Each model uses NumPy to compute lighting across entire meshes at once.

- **08 - Texture Mapping**
  Wrap planets in realistic textures, like Jupiter’s bands or Saturn’s rings. We’ll map 2D images onto 3D surfaces using NumPy arrays, inspired by Blinn’s texture-mapped spheres.

- **09 - Triangles and Watertight Rasterization**
  Render triangle meshes (e.g., for Voyager’s spacecraft) with watertight rasterization. We’ll use vectorized operations to evaluate edge functions and assign colors, ensuring pixel-perfect results.

- **10 - Stars**
  Create a starry backdrop for our animations. We’ll use NumPy to scatter points in 3D space, projecting them as glowing dots to evoke the cosmic vastness of Voyager’s journey.
